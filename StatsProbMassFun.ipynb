```python
import pandas as pd
import numpy as np
```

```python
import random
```

```python
L=[]
for i in range(10000):
    L.append(random.randint(1,6))
  
```

```python
len(L)
```

```python
L[:5]
```

```python
# probabilities for each outcomes value/total sum values
a=pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()
a
```

```python
s = (pd.Series(L).value_counts() / pd.Series(L).value_counts().sum()).sort_index()
```

```python
s
```

```python
s.plot(kind='bar')
```

```python
#if we roll together 2 dice then outcome is
L2=[]
for i in range(10000):
    p=random.randint(1,6)
    k=random.randint(1,6)
    L2.append(p+k)
```

```python
len(L2)
```

```python
L2[:5]
```

```python
s1=(pd.Series(L2).value_counts()/pd.Series(L2).value_counts().sum()).sort_index()
```

```python
s1
```

```python
s1.plot(kind='bar')
```

```python
# cumulative probability fuction for prob.Mass function
r=np.cumsum(s1)
```

```python
r.plot(kind='bar')
```
