```python
import numpy as np
import time 
import sys
```

```python
#list of 100 elemets created 0 to 99
lista=range(100)
lista
```

```python
#array of 100 elemets created 0 to 99
numpyarray=np.arange(100)
numpyarray
```

```python
#      1     Code to find which use less memory list or narray

#sys is used to get the getsize varible to find the size of the list 
p=sys.getsizeof(7)*len(lista)
print('list size in Bytes',p)
#here 7 is the item of the list each item have same or equal space so we multipy it with the length of list to get size of whole list
# size of narray
q=numpyarray.itemsize*numpyarray.size
print("Array size in Bytes",q)
if (p>q):
    print("numpy array take less memmory space then list .Hence proved numpyaaaray is better")
else:
    print("numpy array take more memmory space then list.Hence proved list is better ")
```
