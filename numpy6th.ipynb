Fancy Indexing

```python
import numpy as np
arr1=np.arange(24).reshape(6,4)
arr1
```

```python
#particular row should print
arr1[[1,3,5]]
```

```python
arr2=np.random.randint(low=1,high=91,size=49).reshape(7,7)
```

```python
arr2
```

```python
[arr2>21]
```

```python
arr2[(arr2>21) & (arr2%2!=0)]
```

```python
arr2[(arr2>21)&(arr2%2!=0)]=0
```

```python
arr2
```
