```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import normal
import seaborn as sns
```

PARAMETRIC DISTRIBUTION 

```python
sample=normal(loc=50,scale=5,size=1000)
# loc=mean,scale=standard devition
```

```python
sample
```

```python
sample.mean()
```

```python
sample.std()
```

```python
plt.hist(sample,bins=10)
```

```python
#calculate the sample mean and sample standard deviation
sample_mean=sample.mean()
sample_std=sample.std()
```

```python
#fit the distribution to the above parameter
from scipy.stats import norm
#create normal distribution with sample mean,sample std parameters
distribution=norm(sample_mean,sample_std)
```

```python
a=sample.min()
```

```python
b=sample.max()
```

```python
#generate data points between min and max value
values=np.linspace(a,b,100)
```

```python
#pdf function used to calculate probability density function
probDensity=[distribution.pdf(value)for value in values]
```

```python
plt.hist(sample,bins=10,density=True)
plt.plot(values,probDensity)
```

```python
import seaborn as sns
sns.distplot(sample)
```

```python
sns.histplot(sample, kde=True)
```

NON PARAMETRIC (KDE)

```python
sample1=normal(loc=20,scale=5,size=300)
sample2=normal(loc=40,scale=5,size=700)
sampleT=np.hstack((sample1,sample2))
```

```python
sampleT
```

```python
#plot histogram
plt.hist(sampleT,bins=43)
```

```python
from sklearn.neighbors import KernelDensity
model=KernelDensity(bandwidth=3,kernel='gaussian')
#take data 1d to 2d because ml alogo works on 2d
# convert to 2d array
sampleT=sampleT.reshape(len(sampleT),1)
# model fits means its calculated the y valuse 
# by adding the gaussian values comes to that point above 
model.fit(sampleT)
```

```python
#nowe generate min to max values or data point using linspace
values=np.linspace(sampleT.min(),sampleT.max(),100)
values=values.reshape((len(values),1))
```

Log density estimations

```python
#score_samples()is the log density estimation function which return log values 
probs=model.score_samples(values)
# to cancel the log from the result us exponential function
probs=np.exp(probs)
```

```python
plt.hist(sampleT,bins=50,density=True)
plt.plot(values[:],probs)
plt.show()
```
